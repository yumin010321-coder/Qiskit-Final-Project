```python
import numpy as np

def classical_nash_equilibria(payoff_A, payoff_B):
    equilibria = []

    for i in range(2):      # A strategy: 0=C, 1=D
        for j in range(2):  # B strategy: 0=C, 1=D
            a_payoff = payoff_A[i, j]
            b_payoff = payoff_B[i, j]

            a_best = a_payoff == np.max(payoff_A[:, j])
            b_best = b_payoff == np.max(payoff_B[i, :])

            if a_best and b_best:
                equilibria.append((i, j))

    return equilibria
```

```python
import numpy as np

def U_strategy(theta, phi):
    return np.array([
        [np.exp(1j * phi) * np.cos(theta / 2), np.sin(theta / 2)],
        [-np.sin(theta / 2), np.exp(-1j * phi) * np.cos(theta / 2)]
    ], dtype=complex)
```

```python
def J_gate(gamma):
    I = np.eye(2, dtype=complex)
    D = U_strategy(np.pi, 0)

    return (
        np.cos(gamma / 2) * np.kron(I, I)
        + 1j * np.sin(gamma / 2) * np.kron(D, D)
    )
```

```python
def quantum_game_probabilities(theta_A, phi_A, theta_B, phi_B, gamma):
    initial_state = np.array([1, 0, 0, 0], dtype=complex)  # |00>

    J = J_gate(gamma)
    J_dagger = J.conj().T

    U_A = U_strategy(theta_A, phi_A)
    U_B = U_strategy(theta_B, phi_B)

    final_state = J_dagger @ np.kron(U_A, U_B) @ J @ initial_state

    probs = np.abs(final_state) ** 2

    return {
        "CC": probs[0],
        "CD": probs[1],
        "DC": probs[2],
        "DD": probs[3]
    }
```

```python
def expected_payoff(probs, payoff_A, payoff_B):
    outcomes = ["CC", "CD", "DC", "DD"]
    indices = {
        "CC": (0, 0),
        "CD": (0, 1),
        "DC": (1, 0),
        "DD": (1, 1)
    }

    EA = 0
    EB = 0

    for outcome in outcomes:
        i, j = indices[outcome]
        EA += probs[outcome] * payoff_A[i, j]
        EB += probs[outcome] * payoff_B[i, j]

    return EA, EB
```
